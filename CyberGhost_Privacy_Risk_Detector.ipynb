## Section 1 — Project README

# CyberGhost — Hidden Privacy Risk Detector

**IBM SkillsBuild Cybersecurity BootCamp Project**

---

## Abstract

CyberGhost is a browser-based privacy risk detection tool that scans text for potentially sensitive information — email addresses, phone numbers, API secrets, government-style identifier patterns, and postal address patterns — entirely on the user's device without uploading any data to a server.

---

## Problem Statement

People routinely expose personal and sensitive information when composing social-media posts, emails, shared documents, and public messages. Common accidental disclosures include personal email addresses, mobile phone numbers, embedded API keys or passwords, government-issued identifier numbers, and home addresses. Most users lack a simple, trustworthy tool that can review text locally before sharing.

---

## Objectives

1. Detect seven categories of potentially sensitive text patterns using client-side JavaScript.
2. Calculate an indicative privacy score (0–100) to communicate risk level.
3. Allow users to mask detected information with safe placeholders before sharing.
4. Generate a JSON security report containing only metadata — never the original input.
5. Maintain a strict privacy-by-design architecture: no backend, no uploads, no tracking.

---

## Features

| # | Feature | Description |
|---|---------|-------------|
| 1 | **Text Scanner** | Accepts up to 100,000 characters; validates input; Ctrl+Enter shortcut |
| 2 | **Email Detection** | RFC-ish regex detection of common email formats |
| 3 | **Phone Detection** | Indian mobile (10-digit, +91 prefix), international, US formats |
| 4 | **Secret Detection** | password=, api_key=, token=, Bearer headers, PEM markers |
| 5 | **Identifier Detection** | Aadhaar-like 12-digit, PAN-like, SSN-like patterns |
| 6 | **Address Detection** | Keyword + number heuristics for street/postal addresses |
| 7 | **Privacy Score** | 0–100 score with risk category and deduction explanation |
| 8 | **Automatic Masking** | Replaces findings with [EMAIL REDACTED], [SECRET REDACTED] etc. |
| 9 | **Security Report** | Downloadable JSON with metadata only — no original text |

```

│
```

---

## Detection Limitations

| Limitation | Detail |
|------------|--------|
| Pattern-only matching | Cannot confirm whether a matched value is real or active |
| False positives | Numbers may match phone patterns; addresses rely on keywords |
| Novel formats | Secrets with unusual naming conventions may be missed |
| Encoding | Obfuscated or encoded secrets are not decoded |
| Language | Optimised for English-language text |
| Context-free | Cannot infer whether sharing is safe from context |

Always review findings manually. Do not rely solely on CyberGhost for security decisions.

---

## Privacy-by-Design

- ✅ **No server** — no backend exists
- ✅ **No uploads** — text never leaves the browser tab
- ✅ **No storage** — no localStorage, sessionStorage, or cookies used for scanned content
- ✅ **No analytics** — no tracking scripts or third-party requests
- ✅ **No external resources** — no CDN fonts, external CSS, or remote scripts
- ✅ **XSS safe** — user input is rendered with `textContent`, never `innerHTML`
- ✅ **Report safety** — downloaded JSON never contains the original input text

## Known Limitations

- The phone regex may flag long numeric sequences that are not phone numbers.
- Address detection is keyword-dependent and may miss unusual formats.
- Very long inputs (close to 100,000 chars) are supported but may take slightly longer to scan.
- The Aadhaar regex matches any 12-digit sequence and is not a validator.

```javascript
/**
 * CyberGhost — Hidden Privacy Risk Detector
 * Detection Engine & UI Logic
 *
 * PRIVACY NOTICE:
 * All scanning is performed locally in this browser.
 * No user text is transmitted to any server or external API.
 * No data is stored in localStorage or sessionStorage.
 * No analytics or tracking is performed.
 */

'use strict';

/* =========================================================
   CONSTANTS & CONFIGURATION
   ========================================================= */

const MAX_INPUT_CHARS = 100_000;

const SCORE_WEIGHTS = {
  email:      15,
  phone:      20,
  secret:     35,
  identifier: 25,
  address:    20,
};

const RISK_LEVELS = {
  email:      'Medium',
  phone:      'Medium',
  secret:     'High',
  identifier: 'High',
  address:    'Medium',
};

const CATEGORY_LABELS = {
  email:      'Email Address',
  phone:      'Phone Number',
  secret:     'Secret / Credential',
  identifier: 'Possible Identifier',
  address:    'Possible Address',
};

const MASK_LABELS = {
  email:      '[EMAIL REDACTED]',
  phone:      '[PHONE REDACTED]',
  secret:     '[SECRET REDACTED]',
  identifier: '[IDENTIFIER REDACTED]',
  address:    '[ADDRESS REDACTED]',
};

const PRIORITY = { secret: 0, identifier: 1, address: 2, phone: 3, email: 4 };

/* =========================================================
   DETECTION PATTERNS
   ========================================================= */

const PATTERNS = {
  /**
   * Email – RFC-ish common format. Not a validator, just a detector.
   * Matches: user@domain.tld, user.name+tag@sub.domain.co.in
   */
  email: /\b[a-zA-Z0-9._%+\-]{1,64}@[a-zA-Z0-9.\-]{1,253}\.[a-zA-Z]{2,10}\b/g,

  /**
   * Phone – supports Indian mobile, international +XX formats, US style.
   * Excludes short numbers (< 7 digits total stripped) and obvious years.
   */
  phone: /(?<!\d)(\+?\d{1,3}[\s\-]?)?(\(?\d{2,4}\)?[\s\-]?)?\d{3,5}[\s\-]?\d{4,6}(?!\d)/g,

  /**
   * Secrets – keyword=value assignments and bearer tokens.
   */
  secret: {
    keyValue: /(?:password|passwd|pwd|secret|token|api[_\-]?key|access[_\-]?key|client[_\-]?secret|auth[_\-]?token|bearer|private[_\-]?key|encryption[_\-]?key)\s*[=:]\s*\S+/gi,
    bearerHeader: /Authorization:\s*Bearer\s+\S+/gi,
    privateKeyMarker: /-----BEGIN\s+(?:RSA\s+)?PRIVATE\s+KEY-----/gi,
  },

  /**
   * Possible identifier – Aadhaar-like (12 digits), PAN-like, SSN-like.
   * Described as "possible" patterns, not validated real numbers.
   */
  identifier: {
    aadhaar: /\b\d{4}\s?\d{4}\s?\d{4}\b/g,
    pan:     /\b[A-Z]{5}[0-9]{4}[A-Z]\b/g,
    ssn:     /\b\d{3}[-\s]?\d{2}[-\s]?\d{4}\b/g,
  },

  /**
   * Possible address – keyword anchor + nearby digits or common address tokens.
   */
  address: /(?:(?:\d{1,5}\s+)?(?:[A-Za-z]+\s+){1,4}(?:Street|St|Avenue|Ave|Road|Rd|Lane|Ln|Drive|Dr|Boulevard|Blvd|Way|Court|Ct|Circle|Cir|Place|Pl|Nagar|Colony|Sector|Phase|Layout|Extension|Block|Apartment|Apt|Flat|Floor|Building|Bldg|Tower|Wing|Plot|House|Villa|Gate|Cross|Main)[\s,]?\d{0,5})/gi,
};

/* =========================================================
   FINDING FACTORY
   ========================================================= */

let _findingIdCounter = 0;
function createFinding({ category, start, end, originalText, explanation, recommendation }) {
  const masked = MASK_LABELS[category];
  return {
    id:             `f${++_findingIdCounter}`,
    category,
    severity:       RISK_LEVELS[category],
    label:          CATEGORY_LABELS[category],
    start,
    end,
    originalLength: end - start,
    maskedValue:    masked,
    // Store a brief context snippet (never the full secret value)
    contextSnippet: safeContextSnippet(originalText, start, end, category),
    explanation,
    recommendation,
  };
}

/**
 * Returns a context snippet for display. For secrets, only shows type prefix.
 */
function safeContextSnippet(text, start, end, category) {
  if (category === 'secret') {
    // Only reveal the keyword part, not the actual value
    const raw = text.slice(start, end);
    const eqIdx = raw.search(/[=:]/);
    if (eqIdx !== -1) {
      return raw.slice(0, eqIdx + 1) + '***';
    }
    return '***';
  }
  // For other categories, show the match (it's already in the text)
  const snippet = text.slice(start, end);
  return snippet.length > 60 ? snippet.slice(0, 57) + '…' : snippet;
}

/* =========================================================
   DETECTION FUNCTIONS
   ========================================================= */

/**
 * Detect email addresses.
 * @param {string} text
 * @returns {Array}
 */
function detectEmails(text) {
  const findings = [];
  const rx = new RegExp(PATTERNS.email.source, 'g');
  let m;
  while ((m = rx.exec(text)) !== null) {
    findings.push(createFinding({
      category: 'email',
      start: m.index,
      end: m.index + m[0].length,
      originalText: text,
      explanation:
        'An email address can reveal personal contact information and may be harvested by automated crawlers, sold to marketers, or targeted in phishing attacks.',
      recommendation:
        'Remove the email address or replace it with a general contact form URL if public sharing is intended.',
    }));
  }
  return findings;
}

/**
 * Detect phone numbers.
 * Applies heuristics to filter obvious false positives (dates, short numbers, lone years).
 * @param {string} text
 * @returns {Array}
 */
function detectPhones(text) {
  const findings = [];
  const rx = new RegExp(PATTERNS.phone.source, 'g');
  let m;
  while ((m = rx.exec(text)) !== null) {
    const raw = m[0].trim();
    const digits = raw.replace(/\D/g, '');

    // Filter: too short (less than 7 digits) or too long (more than 15)
    if (digits.length < 7 || digits.length > 15) continue;

    // Filter: looks like a standalone year (4 digits 1900–2099)
    if (/^\d{4}$/.test(digits) && +digits >= 1900 && +digits <= 2099) continue;

    // Filter: looks like a date format (DD/MM/YYYY, MM/DD/YYYY, YYYY-MM-DD)
    if (/^\d{1,4}[\/\-]\d{1,2}[\/\-]\d{2,4}$/.test(raw)) continue;

    // Filter: all same digit (1111111111)
    if (/^(.)\1+$/.test(digits)) continue;

    findings.push(createFinding({
      category: 'phone',
      start: m.index,
      end: m.index + m[0].length,
      originalText: text,
      explanation:
        'This pattern resembles a phone number. Phone numbers can be used for unsolicited contact, SIM-swap attacks, or identity correlation. Pattern matching cannot confirm whether this number is active or belongs to any individual.',
      recommendation:
        'Remove or mask the phone number before sharing this text publicly.',
    }));
  }
  return findings;
}

/**
 * Detect secrets: password assignments, API keys, tokens, bearer headers, PEM markers.
 * @param {string} text
 * @returns {Array}
 */
function detectSecrets(text) {
  const findings = [];

  const secretPatterns = [
    PATTERNS.secret.keyValue,
    PATTERNS.secret.bearerHeader,
    PATTERNS.secret.privateKeyMarker,
  ];

  for (const pat of secretPatterns) {
    const rx = new RegExp(pat.source, pat.flags);
    let m;
    while ((m = rx.exec(text)) !== null) {
      findings.push(createFinding({
        category: 'secret',
        start: m.index,
        end: m.index + m[0].length,
        originalText: text,
        explanation:
          'A potential credential or secret key was detected. Exposed credentials can allow unauthorized access to accounts, APIs, or services. Detection covers common assignment patterns and cannot identify every possible secret format.',
        recommendation:
          'Immediately remove any real credentials from this text. Rotate any exposed credentials. Use environment variables or secret management services instead of embedding secrets in plaintext.',
      }));
    }
  }

  return findings;
}

/**
 * Detect possible government-style identifier patterns.
 * Results are described as possible patterns, never confirmed IDs.
 * @param {string} text
 * @returns {Array}
 */
function detectPossibleIdentifiers(text) {
  const findings = [];

  const idPatterns = [
    { rx: PATTERNS.identifier.aadhaar, name: 'Aadhaar-like (12-digit)' },
    { rx: PATTERNS.identifier.pan,     name: 'PAN-like' },
    { rx: PATTERNS.identifier.ssn,     name: 'SSN-like' },
  ];

  for (const { rx, name } of idPatterns) {
    const r = new RegExp(rx.source, rx.flags);
    let m;
    while ((m = r.exec(text)) !== null) {
      findings.push(createFinding({
        category: 'identifier',
        start: m.index,
        end: m.index + m[0].length,
        originalText: text,
        explanation:
          `This matches a possible ${name} identifier pattern. This tool cannot confirm that the number is a valid or genuine government-issued ID or that it belongs to any specific person. It is flagged as a precaution.`,
        recommendation:
          'Remove or mask any government-issued identification numbers from text you intend to share publicly. Exposing ID numbers can facilitate identity theft.',
      }));
    }
  }

  return findings;
}

/**
 * Detect possible postal/home address patterns.
 * Detection is approximate and may produce false positives.
 * @param {string} text
 * @returns {Array}
 */
function detectPossibleAddresses(text) {
  const findings = [];
  const rx = new RegExp(PATTERNS.address.source, PATTERNS.address.flags);
  let m;
  while ((m = rx.exec(text)) !== null) {
    const raw = m[0].trim();
    if (raw.split(/\s+/).length < 2) continue; // too short
    findings.push(createFinding({
      category: 'address',
      start: m.index,
      end: m.index + m[0].length,
      originalText: text,
      explanation:
        'This pattern resembles a postal or street address. Address detection is approximate and may produce false positives. Sharing a home address publicly can enable physical stalking, unsolicited mail, or targeted attacks.',
      recommendation:
        'Remove or generalise any home or personal address before sharing this text publicly. Use only city or region-level location when precise location is not necessary.',
    }));
  }
  return findings;
}

/* =========================================================
   OVERLAP RESOLUTION
   ========================================================= */

/**
 * Removes overlapping findings by keeping the highest-priority finding
 * for any overlapping range. Also deduplicates identical offsets.
 *
 * Priority (lower number = higher priority): secret > identifier > address > phone > email
 *
 * @param {Array} findings
 * @returns {Array} Non-overlapping findings sorted by start offset
 */
function resolveOverlaps(findings) {
  if (findings.length === 0) return [];

  // Sort by priority first, then by start position
  const sorted = [...findings].sort((a, b) => {
    const pd = (PRIORITY[a.category] ?? 99) - (PRIORITY[b.category] ?? 99);
    if (pd !== 0) return pd;
    return a.start - b.start;
  });

  const accepted = [];
  for (const f of sorted) {
    const overlaps = accepted.some(a => f.start < a.end && f.end > a.start);
    if (!overlaps) accepted.push(f);
  }

  // Return in document order
  return accepted.sort((a, b) => a.start - b.start);
}

/**
 * Alias for resolveOverlaps to conform to specification naming.
 * Normalizes and resolves overlapping/adjacent findings.
 */
function normalizeFindings(findings) {
  return resolveOverlaps(findings);
}

/* =========================================================
   PRIVACY SCORE
   ========================================================= */

/**
 * Calculates indicative privacy score (0–100).
 * Counts unique occurrences per category, not per duplicate.
 *
 * @param {Array} findings Non-overlapping deduplicated findings
 * @returns {{ score: number, categoryCounts: Object, deductions: Object }}
 */
function calculatePrivacyScore(findings) {
  const categoryCounts = {};
  const deductions = {};

  for (const f of findings) {
    categoryCounts[f.category] = (categoryCounts[f.category] || 0) + 1;
  }

  let totalDeduction = 0;
  for (const [cat, count] of Object.entries(categoryCounts)) {
    const weight = SCORE_WEIGHTS[cat] ?? 10;
    // First occurrence gets full weight; subsequent occurrences get half weight
    const deduction = weight + (count - 1) * Math.ceil(weight / 2);
    deductions[cat] = deduction;
    totalDeduction += deduction;
  }

  const score = Math.max(0, Math.min(100, 100 - totalDeduction));
  return { score, categoryCounts, deductions };
}

/* =========================================================
   MASKING
   ========================================================= */

/**
 * Replaces detected ranges in the original text with category-specific placeholders.
 * Handles multiple non-overlapping findings correctly using offset adjustment.
 *
 * @param {string} text Original input text
 * @param {Array} findings Sorted, non-overlapping findings
 * @returns {string} Masked text
 */
function maskFindings(text, findings) {
  if (findings.length === 0) return text;

  let result = '';
  let cursor = 0;

  for (const f of findings) {
    // Append text before this finding
    result += text.slice(cursor, f.start);
    // Append the mask placeholder
    result += f.maskedValue;
    cursor = f.end;
  }

  // Append remaining text after last finding
  result += text.slice(cursor);
  return result;
}
```

## Section 5 — Detection Logic in Python Pseudocode

The cell below contains a runnable Python equivalent of the core detection logic from `script.js`. It uses only the Python standard library — no packages required. Run it to see a demonstration with fictional sample data.


````python
"""
CyberGhost — Detection Logic (Python pseudocode / conceptual equivalent)

This code documents the logic implemented in script.js.
The browser application runs the equivalent logic in JavaScript — not Python.
No real credentials or personal data appear in this file.
"""

import re

SCORE_WEIGHTS = {
    "email":      15,
    "phone":      20,
    "secret":     35,
    "identifier": 25,
    "address":    20,
}

MASK_LABELS = {
    "email":      "[EMAIL REDACTED]",
    "phone":      "[PHONE REDACTED]",
    "secret":     "[SECRET REDACTED]",
    "identifier": "[IDENTIFIER REDACTED]",
    "address":    "[ADDRESS REDACTED]",
}

PRIORITY = {"secret": 0, "identifier": 1, "address": 2, "phone": 3, "email": 4}


def detect_emails(text):
    pattern = r"[a-zA-Z0-9._%+\-]{1,64}@[a-zA-Z0-9.\-]{1,253}\.[a-zA-Z]{2,10}"
    return [(m.start(), m.end(), "email") for m in re.finditer(pattern, text)]


def detect_phones(text):
    pattern = r"(?<!\d)(\+?\d{1,3}[\s\-])?(\(?\d{2,4}\)?[\s\-])?\d{3,5}[\s\-]?\d{4,6}(?!\d)"
    results = []
    for m in re.finditer(pattern, text):
        digits = re.sub(r"\D", "", m.group())
        if len(digits) < 7 or len(digits) > 15:
            continue
        if re.match(r"^(.)\1+$", digits):
            continue  # all same digit
        if len(digits) == 4 and 1900 <= int(digits) <= 2099:
            continue  # year-like
        results.append((m.start(), m.end(), "phone"))
    return results


def detect_secrets(text):
    pattern = (r"(?:password|passwd|pwd|secret|token|api[_\-]?key|"
               r"access[_\-]?key|client[_\-]?secret|auth[_\-]?token|"
               r"bearer|private[_\-]?key)[\s]*[=:][\s]*\S+")
    return [(m.start(), m.end(), "secret") for m in re.finditer(pattern, text, re.IGNORECASE)]


def detect_possible_identifiers(text):
    results = []
    for pat in [
        r"\b\d{4}\s?\d{4}\s?\d{4}\b",   # Aadhaar-like
        r"\b[A-Z]{5}[0-9]{4}[A-Z]\b",          # PAN-like
        r"\b\d{3}[\-\s]?\d{2}[\-\s]?\d{4}\b", # SSN-like
    ]:
        results += [(m.start(), m.end(), "identifier") for m in re.finditer(pat, text)]
    return results


def detect_possible_addresses(text):
    pattern = (r"(?:(?:\d{1,5}\s+)?(?:[A-Za-z]+\s+){1,4}"
               r"(?:Street|St|Avenue|Ave|Road|Rd|Lane|Ln|Drive|Dr|"
               r"Boulevard|Blvd|Nagar|Colony|Sector|Block|Apartment|Apt|"
               r"Flat|Floor|Building|Plot|House|Cross|Main)[\s,]?\d{0,5})")
    return [(m.start(), m.end(), "address") for m in re.finditer(pattern, text, re.IGNORECASE)]


def resolve_overlaps(findings):
    sorted_f = sorted(findings, key=lambda x: (PRIORITY.get(x[2], 99), x[0]))
    accepted = []
    for f in sorted_f:
        overlaps = any(f[0] < a[1] and f[1] > a[0] for a in accepted)
        if not overlaps:
            accepted.append(f)
    return sorted(accepted, key=lambda x: x[0])


def calculate_privacy_score(findings):
    counts = {}
    for (_, _, cat) in findings:
        counts[cat] = counts.get(cat, 0) + 1
    total_deduction = 0
    for cat, count in counts.items():
        w = SCORE_WEIGHTS.get(cat, 10)
        total_deduction += w + (count - 1) * ((w + 1) // 2)
    return max(0, min(100, 100 - total_deduction))


def mask_findings(text, findings):
    result, cursor = "", 0
    for (start, end, cat) in sorted(findings, key=lambda x: x[0]):
        result += text[cursor:start]
        result += MASK_LABELS.get(cat, "[REDACTED]")
        cursor = end
    return result + text[cursor:]


def generate_safe_report(findings, score):
    """Returns a dict that NEVER includes original text or full secret values."""
    return {
        "privacyScore": score,
        "totalFindings": len(findings),
        "findings": [
            {"category": cat, "placeholder": MASK_LABELS.get(cat, "[REDACTED]")}
            for (_, _, cat) in findings
        ],
        "disclaimer": (
            "Educational indicator only. Not a security certification. "
            "Pattern matching may produce false positives."
        ),
    }


# ---- Demo run (fictional data only) ----
if __name__ == "__main__":
    sample = (
        "Contact arjun@example.com or call +91 9876543210. "
        "api_key=demo_fake_key_abc123"
    )
    all_findings = (
        detect_emails(sample)
        + detect_phones(sample)
        + detect_secrets(sample)
        + detect_possible_identifiers(sample)
        + detect_possible_addresses(sample)
    )
    resolved = resolve_overlaps(all_findings)
    score    = calculate_privacy_score(resolved)
    masked   = mask_findings(sample, resolved)
    report   = generate_safe_report(resolved, score)

    print(f"Findings : {len(resolved)}")
    print(f"Score    : {score}/100")
    print(f"Masked   : {masked}")
    print("Report   :", json.dumps(report, indent=2))
````

## Section 6 — Privacy and Security Notes

### Privacy Architecture
- All detection runs in the browser (JavaScript). No text is uploaded anywhere.
- The downloaded JSON report contains only metadata — never the original input.
- No localStorage, sessionStorage, or cookies store scanned content.
- No external scripts, CDN fonts, or analytics are loaded.

### XSS Prevention
- User input is rendered using `element.textContent` (not `innerHTML`).
- DOM nodes are built with `createElement` / `appendChild`.

### Limitations
- Pattern matching cannot detect every format of sensitive data.
- False positives may occur (e.g., numeric sequences flagged as phone numbers).
- The privacy score is an educational indicator, not a security certification.
